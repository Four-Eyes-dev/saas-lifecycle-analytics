```python
# Do activated users behave meaningfully differently?
# This is what  I had to do 

# Merge activation flag back into events
events_with_activation = events.merge(
    clean_first_event[['user_id', 'activated']],
    on='user_id',
    how='left'
)

# Events per user by activation status
events_per_user = (
    events_with_activation
    .groupby(['user_id', 'activated'])
    .size()
    .reset_index(name='event_count')
)

events_per_user.groupby('activated')['event_count'].describe()

# MEANING OF THE RESULTS BELOW

# ______Activated users_______:

#Perform more actions overall

#Have a higher engagement ceiling (max = 13 vs 10)

# ______Non-activated users_______:

#Show shallower engagement

#Likely drop off earlier in the journey

#_____Key takeaway_______;
#Activation is not cosmetic — it reflects real behavioral depth.

#This validates my activation definition 
```

```python
# DOES ACTIVATION REDUCE CHURN?

# Merge activation into subscriptions
subs_with_activation = subscriptions.merge(
    clean_first_event[['user_id', 'activated']],
    on='user_id',
    how='left'
)

# Create churn flag
subs_with_activation['churned'] = subs_with_activation['churn_date'].notna().astype(int)

# Churn rate by activation status
churn_rate = (
    subs_with_activation
    .groupby('activated')['churned']
    .mean()
    .reset_index()
)

churn_rate

# MEANING OF THE RESULTS SHOWN

# - Every churned user came from the non-activated group

# - Activated users had zero churn

# - Activation is a leading indicator of retention
```

```python
# Final visuals to explain the project better

# EVENT COUNT DISTRIBUTION
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(
    data=events_per_user,
    x='activated',
    y='event_count'
)
plt.title('Event Volume by Activation Status')
plt.show()

# CHURN RATE BAR CHART
sns.barplot(
    data=churn_rate,
    x='activated',
    y='churned'
)
plt.title('Churn Rate by Activation Status')
plt.show()

# In simple terms from the graphs results, 
#Users who fail to activate are significantly more likely to churn, while activated users show stronger engagement and full retention.
```

```python
# _____Key Insights______

# - Activated users generated more product events and showed deeper engagement.

# - Non-activated users had a 15% churn rate, while activated users had zero churn.

# - Activation is a strong early signal of long-term retention.

# - Growth & Lifecycle Implications

# - Early lifecycle campaigns should focus on driving meaningful actions within the first few days.

# - Non-activated users should be targeted with onboarding nudges, education, or assisted activation.

# - Activation metrics can be used to prioritize sales or customer success outreach.
```
